```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
import numpy as np
from google.colab import files

# Upload dataset
uploaded = files.upload()

# Load dataset
file_name = list(uploaded.keys())[0]
data = pd.read_csv(file_name)

# Handle non-numeric data by converting to numeric representations
for column in ['Real Answer', 'Answer', 'Words', 'Unique_Words', 'Stop_Words', 'Non_Stop_Words',
          'Sentences', 'Long_Words', 'Word_Frequencies', 'Real Answer Embedding',
          'Answer Embedding', 'Rating Normalized', 'Adjusted Similarity']:
    data[column] = data[column].astype('category').cat.codes

# Select features and target variable
X = data[['Real Answer', 'Answer', 'Words', 'Unique_Words', 'Stop_Words', 'Non_Stop_Words',
          'Sentences', 'Long_Words', 'Word_Frequencies', 'Real Answer Embedding',
          'Answer Embedding', 'Rating Normalized', 'Adjusted Similarity']]
y = data['Rating']

# Split data into training/validation and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=20/170, random_state=42)

# Scale the data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Build and train the logistic regression model
model = LogisticRegression(max_iter=1000)
model.fit(X_train_scaled, y_train)

# Predict and evaluate
y_pred = model.predict(X_test_scaled)

accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
print("Classification Report:\n", classification_report(y_test, y_pred))

# Output actual vs predicted values to CSV
output_df = pd.DataFrame({
    'Actual Rating': y_test.values,
    'Predicted Rating': y_pred
})
output_df.to_csv('predicted_ratings.csv', index=False)
print("Output CSV generated: predicted_ratings.csv")

# Make CSV downloadable
files.download('predicted_ratings.csv')

# Predict for new unseen data (20 rows)
unseen_data = X_test_scaled[:20]  # Assuming the last 20 rows are unseen
unseen_predictions = model.predict(unseen_data)
print("Predictions for unseen data:", unseen_predictions)
```
