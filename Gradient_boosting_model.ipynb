```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score, classification_report
from sklearn.preprocessing import LabelEncoder
from xgboost import XGBClassifier
from google.colab import files
from sklearn.ensemble import GradientBoostingClassifier

# Upload the CSV file
uploaded = files.upload()

# Load the CSV file
data = pd.read_csv('grader.csv')


```

```python
# Ensure the dataset contains the required columns
assert 'Answer' in data.columns, "Column 'Answer' is missing."
assert 'Real Answer' in data.columns, "Column 'Real Answer' is missing."

# Split the data into training/testing and unseen datasets
train_test_data = data[:150]
unseen_data = data[150:]

# Define features (X) and target (y)
X = train_test_data['Answer']
y = train_test_data['Real Answer']


# Preprocess the text data using TfidfVectorizer
tfidf = TfidfVectorizer()
X_transformed = tfidf.fit_transform(X)

# Encode the target labels
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

# Split the training and testing data
X_train, X_test, y_train, y_test = train_test_split(X_transformed, y_encoded, test_size=0.2, random_state=42)

# Convert sparse matrix to dense (if necessary)
X_train = X_train.toarray()
X_test = X_test.toarray()

# Initialize and train the Gradient Boosting model (XGBoost)
model = XGBClassifier(use_label_encoder=False, eval_metric='logloss', random_state=42)
model.fit(X_train, y_train)

# Evaluate the model on the test set
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f'Accuracy on test data: {accuracy:.2f}')
print("Classification Report:\n", classification_report(y_test, y_pred))

# Preprocess and predict on unseen data
X_unseen = unseen_data['Answer']
X_unseen_transformed = tfidf.transform(X_unseen).toarray()
y_unseen_pred = model.predict(X_unseen_transformed)
unseen_data['predicted_answers'] = label_encoder.inverse_transform(y_unseen_pred)

# Save predictions for unseen data
unseen_data.to_csv('unseen_data_predictions.csv', index=False)
print("Predictions for unseen data saved to 'unseen_data_predictions.csv'.")
```
